```python
a=int(input("ENTER AN INTEGER= "))
if a % 7 == 0:
    print(a,"GIVEN NUMBER IS DIVISIBLE BY 7")
else:
    print(a,"GIVEN NUMBER IS NOT DIVISIBLE BY 7")
```
